```python
import json
import pandas as pd
from pathlib import Path

EVENTS_PATH = Path("../data/raw/statsbomb/events")

```

```python
event_files = list(EVENTS_PATH.glob("*.json"))

print(f"{len(event_files)} event files found")
event_files[:5]
```

```python
import json

sample_file = event_files[0]

with open(sample_file, "r", encoding="utf-8") as f:
    sample_events = json.load(f)

len(sample_events), sample_events[0]
```

```python
import pandas as pd

df_sample = pd.json_normalize(sample_events)

df_sample.head()
```

```python
match_id = int(sample_file.stem)

df_sample = pd.json_normalize(sample_events)
df_sample["match_id"] = match_id

df_sample.head()
```

```python
useful_columns = [
    "match_id",
    "type.name",
    "minute",
    "second",
    "player.id",
    "player.name",
    "team.name",
    "location",
    "pass.length",
    "shot.statsbomb_xg",
    "dribble.outcome.name"
]

df_sample_light = df_sample[useful_columns]
df_sample_light.head()
```

```python
USEFUL_COLUMNS = [
    "match_id",
    "type.name",
    "minute",
    "second",
    "player.id",
    "player.name",
    "team.name",
    "location",
    "pass.length",
    "shot.statsbomb_xg",
    "dribble.outcome.name"
]
```

```python
all_events = []

for file in event_files:
    match_id = int(file.stem)

    with open(file, "r", encoding="utf-8") as f:
        events = json.load(f)

    df_match = pd.json_normalize(events)
    df_match["match_id"] = match_id

    # garder uniquement les colonnes existantes
    existing_cols = [c for c in USEFUL_COLUMNS if c in df_match.columns]
    df_match_light = df_match[existing_cols]

    all_events.append(df_match_light)

df_events_light = pd.concat(all_events, ignore_index=True)

df_events_light.head()
```

```python
print(df_events_light.shape)
df_events_light["type.name"].value_counts().head(10)
```

```python
from pathlib import Path

output_path = Path("../data/processed/events_light.parquet")

output_path.parent.mkdir(parents=True, exist_ok=True)

df_events_light.to_parquet(output_path, index=False)

print(f"Saved file at: {output_path.resolve()}")
```
